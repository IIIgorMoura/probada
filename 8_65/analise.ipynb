```python
import matplotlib.pyplot as plt

# Exemplo com gráfico de dispersão

# Dados
x1=[1,2.5,2.7,3.3,4.4,5.6,7.7,8,10]
y1=[2.2,3.8,5.6,6.8,7.9,10,11,12.8,13.9]
x2=[100,150,250,270,330,550,700,810]
y2=[150,200,230,530,690,890,1000,1230]
# Cria uma nova janela para plotagem
plt.figure("Dispersão")
# Cria o gráfico de dispersão
#plt.scatter(x1,y1)
# Pode-se inserir duas populações
plt.scatter(x1,y1,color='blue',label='População 1')
plt.scatter(x2,y2,color='red',label='População 2')
# Título e nomes dos eixos
plt.title("Gráfico de dispersão")
plt.xlabel("X")
plt.ylabel("Y")
# Desenha as grades
plt.grid(True)
# Exibe a janela com o gráfico
plt.show()   # se tiver apenas um comando plt.show() exibe as janelas simultaneamente

# Exemplo com gráfico pizza

def valores_pizza(val):   # rotina para mostrar as porcentagens e os valores no gráfico pizza
    total=sum(valores)
    valor=int(round(val*total/100))
    return f'{valor}\n({val:.1f}%)'
    # :.1f indica número decimal com 1 casa após a vírgula/ponto.

# Dados
valores=[10,20,25,35,44]
nomes=["A","B","C","D","E"]
# Cria uma nova janela para plotagem
plt.figure("Pizza")
# Desenha o gráfico pizza
plt.pie(valores,labels=nomes,autopct=valores_pizza)  # mostra porcentagens e valores
# Título do gráfico
plt.title("Gráfico de pizza")
# Exibe a janela com o gráfico
plt.show()

# Exemplo com gráfico de barras

# Cria uma nova janela para plotagem
plt.figure("Barras")
# Desenha o gráfico de barras
plt.bar(nomes,valores)
# Título e nomes nos eixos
plt.title("Gráfico de barras")
plt.xlabel("Categorias")
plt.ylabel("Valores")
# Exibe a janela com o gráfico
plt.show()

# Exemplo com gráfico boxplot

grupo1=[1,2,3,4,5,6,8,10,25]
grupo2=[-15,1,2,5,5,7,22,32]
grupo3=[-20,-5,-2,0,1,2,3,4,20,30]
# Cria uma nova janela para plotagem
plt.figure("Múltiplos Boxplots")
# Desenhando os grupos
plt.boxplot([grupo1,grupo2,grupo3])
# Nomes dos grupos
plt.xticks([1,2,3],["Grupo 1","Grupo 2","Grupo 3"])
# Título do gráfico
plt.title("Comparação entre grupos")
# Exibe a janela com o gráfico
plt.show()

# Exemplo de histograma

# Dados para o histograma
dados=[1,1,1,2,3,3,3,3,5,5,6,7,8,1,2,9,8,7,5,6,10,11,10,10,5,7,1,1,1,2]
# Cria uma nova janela para plotagem
plt.figure("Histograma")
# Desenhando o histograma com 5 intervalos
plt.hist(dados,bins=5)
# Título e nomes dos eixos
plt.title("Histograma")
plt.xlabel("Valores")
plt.ylabel("Frequência")
# Exibe a janela com o gráfico
plt.show()
```
